### t test 예시
**검정: 남학생과 여학생의 점수 차이가 나는가?**

남학생 점수: 70, 75, 80, 65, 90

여학생 점수: 85, 88, 92, 78, 95

```python
import numpy as np
import scipy
```

```python
x1_raw = [70, 75, 80, 65, 90]
x2_raw = [85, 88, 92, 78, 95]
x1_arr = np.array(x1_raw)
x2_arr = np.array(x2_raw)
x1_mean = np.mean(x1_arr)
x2_mean = np.mean(x2_arr)
x1_var = np.var(x1_arr, ddof=1)
x2_var = np.var(x2_arr, ddof=1)
x1_std = np.sqrt(x1_var)
x2_std = np.sqrt(x2_var)
print(f"x1_mean: {x1_mean}, x1_var: {x1_var}, x1_std: {x1_std}")
print(f"x2_mean: {x2_mean}, x2_var: {x2_var}, x2_std: {x2_std}")
print(f"x1_mean-x2_mean: {x1_mean - x2_mean}")
print(f"x1_std-x2_std: {x1_std - x2_std}")
```

```python
# 등분산 가정
pooled_var = ((len(x1_arr) - 1) * x1_var + (len(x2_arr) - 1) * x2_var) / (
    len(x1_arr) + len(x2_arr) - 2
)
pooled_std = np.sqrt(pooled_var)
std_error = pooled_std * np.sqrt(1 / len(x1_arr) + 1 / len(x2_arr))
t_statistic = (x1_mean - x2_mean) / std_error
degrees_of_freedom = len(x1_arr) + len(x2_arr) - 2
p_value = 2 * (1 - scipy.stats.t.cdf(abs(t_statistic), df=degrees_of_freedom))
cohens_d = np.abs((x1_mean - x2_mean) / pooled_std)
ci_95_mean_diff = (
    (x1_mean - x2_mean) - scipy.stats.t.ppf(0.975, df=degrees_of_freedom) * std_error,
    (x1_mean - x2_mean) + scipy.stats.t.ppf(0.975, df=degrees_of_freedom) * std_error,
)
caculated_t_statistic, calculated_p_value = scipy.stats.ttest_ind(
    x1_arr, x2_arr, equal_var=True
)
print(
    f"pooled_var: {pooled_var}, pooled_std: {pooled_std}, std_error: {std_error}, degrees_of_freedom: {degrees_of_freedom}"
)
print(f"t_statistic: {t_statistic}, p_value: {p_value}, cohens_d: {cohens_d}")
print(
    f"calculated_t_statistic: {caculated_t_statistic}, calculated_p_value: {calculated_p_value}"
)
print(f"ci_95_mean_diff: {ci_95_mean_diff}")
```

p value > 0.05이므로 귀무가설을 기각할 수 없다. 그러나, Cohen's d는 1.4로 효과크기는 크기 때문에, 표본이 적어 불확실할 수 있다.
