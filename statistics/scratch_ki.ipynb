```python
# seed 고정
seed = 42
rng = np.random.default_rng(seed)
```

#### 카이제곱 분포를 찾는 통계 모형 따라하기

```python
def draw_histogram(data, title, xlabel, ylabel, bins):
    plt.hist(data, bins=bins)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()
```

```python
# 모집단 선언

parents = rng.standard_normal(10000)
draw_histogram(parents, "Histogram of Parents Population", "Value", "Frequency", 30)
```

```python
sample_size_list = [50, 500, 2000]
```

```python
def get_sample_and_plot(sample_size):
    sample = rng.choice(parents, size=sample_size, replace=False)
    draw_histogram(
        sample, f"Histogram of Sample (n={sample_size})", "Value", "Frequency", 15
    )
    return sample
```

```python
# 표본 추출
result_dict = {}
for n in sample_size_list:
    sample = get_sample_and_plot(n)
    result_dict[n] = sample
```

```python
# 추출한 표본을 모집단의 분산과 비교
for n, sample in result_dict.items():
    sample_variance = np.var(sample, ddof=1)
    population_variance = np.var(parents, ddof=1)
    print(
        f"Sample Size: {n}, Sample Variance: {sample_variance:.4f}, Population Variance: {population_variance:.4f}"
    )
```
