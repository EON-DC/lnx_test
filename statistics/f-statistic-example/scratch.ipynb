예를 들어 세 반의 평균 점수가 다르다고 합시다.

A반: 70, 75, 80

B반: 85, 88, 90

C반: 72, 74, 76


눈으로 보면 B반이 높아 보입니다.

그런데 통계적으로는 이렇게 묻습니다.

“집단 평균 차이가, 집단 내부의 흔들림보다 충분히 큰가?”

이걸 수치화한 것이 F 통계량 입니다.

$F=\frac{집단 간 분산(Mean Square Between)}{집단 내 분산(Mean Square Within)}$


```python
import pandas as pd
import scipy.stats as stats
```

```python
data = [[70, 75, 80], [85, 88, 90], [72, 74, 76]]
df = pd.DataFrame(data, index=["A", "B", "C"])
print(df)
```

```python
# F-통계량 계산
f_statistic, p_value = stats.f_oneway(df.loc["A"], df.loc["B"], df.loc["C"])
print("F-statistic:", f_statistic)
print("p-value:", p_value)

#
```

F가 크면
그룹 평균 차이가 큼
그룹 내부 잡음은 상대적으로 작음
따라서 “집단 차이가 있다”는 쪽으로 감

F가 1에 가까우면
집단 간 차이와 집단 내 흔들림이 비슷함
그룹 차이라고 보기 어려움

F가 매우 작으면
그룹 평균 차이보다 내부 잡음이 더 큼

```python
# 집단간 제곱합(Sum of Squares Between, SSB)
SSB = sum(df.shape[1] * (df.mean(axis=1) - df.values.flatten().mean()) ** 2)
# 집단내 제곱합(Sum of Squares Within, SSW)
SSW = sum(((df - df.mean(axis=1).values.reshape(-1, 1)) ** 2).values.flatten())
# 자유도 계산
df_between = df.shape[0] - 1
df_within = df.shape[0] * (df.shape[1] - 1)
# 평균 제곱 계산
MSB = SSB / df_between
MSW = SSW / df_within
# F-통계량 계산
manual_f_statistic = MSB / MSW
print("Manual F-statistic:", manual_f_statistic)

# p-value 계산
manual_p_value = 1 - stats.f.cdf(manual_f_statistic, df_between, df_within)
print("Manual p-value:", manual_p_value)
```
