합격	불합격	행합

남자	30	20	50

여자	10	40	50

열합	40	60	100

성별: 남 / 여

합격 여부: 합격 / 불합격

독립성 검정의 귀무가설은:

𝐻0:두 변수는 독립이다 (성별을 안다고 해서 합격 여부를 더 잘 알 수 없다.)

H1:두 변수는 독립이 아니다 (두 변수 사이에 연관성이 있다.)

```python
import pandas as pd
import scipy.stats as stats
```

```python
df = pd.DataFrame([[30, 20], [10, 40]])
df.columns = ["합격", "불합격"]
df.index = ["남자", "여자"]
print(df)
```

```python
# 독립이라면 기대값은 어떻게 생겨야 하는가
chi2, p, dof, expected = stats.chi2_contingency(df, correction=False)
print("Chi2:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)
```

```python
# 직접 계산
total = df.values.sum()
e11 = df.iloc[0, :].sum() * df.iloc[:, 0].sum() / total
e12 = df.iloc[0, :].sum() * df.iloc[:, 1].sum() / total
e21 = df.iloc[1, :].sum() * df.iloc[:, 0].sum() / total
e22 = df.iloc[1, :].sum() * df.iloc[:, 1].sum() / total
row_sums = df.sum(axis=1)
col_sums = df.sum(axis=0)
print("Total:", total)
print("Row sums:", row_sums)
print("Column sums:", col_sums)
expected_manual = pd.DataFrame(
    [[e11, e12], [e21, e22]], index=df.index, columns=df.columns
)
print("Expected frequencies (manual calculation):\n", expected_manual)

# 카이제곱 통계량 계산
chi2_manual_df = (df - expected_manual) ** 2 / expected_manual
print("Chi2 components (manual calculation):\n", chi2_manual_df)
chi2_manual = chi2_manual_df.values.sum()
print("Chi2 (manual calculation):", chi2_manual)
# 독립성 검정이므로, df = (행의 수 - 1) * (열의 수 - 1) = 1
dof_manual = (df.shape[0] - 1) * (df.shape[1] - 1)
print("Degrees of freedom (manual calculation):", dof_manual)
p_manual = 1 - stats.chi2.cdf(chi2_manual, dof_manual)
print("p-value (manual calculation):", p_manual)
print("Chi2 (scipy):", chi2)
print("p-value (scipy):", p)

# Cramér's V 계산
n = total
k = min(df.shape)  # 행과 열의 수 중 작은 값
cramers_v = (chi2 / (n * (k - 1))) ** 0.5
print("Cramér's V:", cramers_v)
```
