```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
cols = ["height", "weight"]
mean_height = 170
std_height = 13

mean_weight = 80
std_weight = 10

n_size = 100
random_seed = 1618

# set random seed
np.random.seed(random_seed)

height_data = np.random.normal(loc=mean_height, scale=std_height, size=n_size)
weight_data = np.random.normal(loc=mean_weight, scale=std_weight, size=n_size)
data = pd.DataFrame({"height": height_data, "weight": weight_data})
```

```python
data.describe()
```

```python
# visualize data
def visualize_data(data):
    # 1 scatter, 2 histogram, 3 boxplot
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].scatter(data["height"], data["weight"])
    axes[0].set_xlabel("Height")
    axes[0].set_ylabel("Weight")
    axes[0].set_title("Height vs Weight")
    axes[1].hist(data["height"], bins=20, alpha=0.7, label="Height")
    axes[1].hist(data["weight"], bins=20, alpha=0.7, label="Weight", color="orange")
    axes[1].set_xlabel("Value")
    axes[1].set_ylabel("Frequency")
    axes[1].set_title("Histogram of Height and Weight")
    axes[1].legend()
    axes[2].boxplot([data["height"], data["weight"]], tick_labels=["Height", "Weight"])
    axes[2].set_title("Boxplot of Height and Weight")
    plt.tight_layout()
    plt.show()


visualize_data(data)
```

```python
# add skewness and kurtosis
from scipy.stats import skew, kurtosis

x = np.random.normal(0, 1, 1000)
idx = np.random.choice(len(x), size=300, replace=False)
x[idx] += 15
print("Skewness of x:", skew(x))
print("Kurtosis of x:", kurtosis(x))
```

```python
def visualize_skew_kurtosis(x):
    fig, ax = plt.subplots(1, 2, figsize=(10, 5))
    sns.histplot(x, bins=30, kde=True, ax=ax[0])
    ax[0].set_title("Histogram and KDE of x")
    sns.kdeplot(x=x, ax=ax[1])
    ax[1].set_title("KDE(Kernel Density Estimation) of x")
    plt.tight_layout()
    plt.show()


visualize_skew_kurtosis(x)
```

```python
# where 사용
edges = np.histogram_bin_edges(x, bins=2)
print(edges)
```

```python
# Shapiro-Wilk test
from scipy.stats import shapiro

x = np.random.normal(0, 1, 100)
# x = np.array([12, 13, 14, 15, 15, 16, 17, 18, 19, 20])
stat, p = shapiro(x)

print("W =", stat)
print("p-value =", p)
visualize_skew_kurtosis(x)
```
